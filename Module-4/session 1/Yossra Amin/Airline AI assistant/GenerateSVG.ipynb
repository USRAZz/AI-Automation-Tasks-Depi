````python
challenge = "a panda rollerblading to work"
prompt = f"Generate an SVG of {challenge}. Respond with the SVG only, no code blocks."
messages = [{"role": "user", "content": prompt}]
````

````python
from datetime import datetime

def artist(model, effort=None):
    try:
        start = datetime.now()
        
        # we use our client directed to Groq instead of anthropic
        response = client.chat.completions.create(
            model=model,
            messages=messages
        )
        
        result = response.choices[0].message.content
        end = datetime.now()
        elapsed = (end - start).total_seconds()
        
        heading = f"### {model}\n**Time:** {elapsed // 60:.0f} min {elapsed % 60:.0f} s\n\n"
        
    except Exception as e:
        print(f"Model {model} failed: {e}")
        heading = f"### {model}\n**Error:** {e}\n\n"
        return heading, None
        
    return heading, result
````

````python
heading, result = artist(MODEL)
````

````python
from IPython.display import display, HTML, Markdown

def reveal(svg_text):
    if not svg_text:
        print("لا يوجد كود SVG للعرض!")
        return
        
    # clean the code if markdown is present
    cleaned = svg_text.strip()
    if "```" in cleaned:
        # extract the code between the markdown tags
        parts = cleaned.split("```")
        for part in parts:
            if "<svg" in part:
                cleaned = part
                if cleaned.startswith("xml") or cleaned.startswith("svg"):
                    cleaned = cleaned.split("\n", 1)[1]
                break
                
    # display the title or the clean code as HTML
    display(HTML(cleaned))
````

````python
display(Markdown(heading))
if result:
    reveal(result)
````
