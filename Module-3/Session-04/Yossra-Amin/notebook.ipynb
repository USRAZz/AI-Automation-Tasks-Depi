```python
# Read System_Prompt
with open('system_prompt.txt', 'r', encoding='utf-8') as f:
    system_prompt = f.read()

# Read User_Prompt
with open('user_prompt.txt', 'r', encoding='utf-8') as f:
    user_prompt_template = f.read()

print("System Prompt and User Prompt loaded successfully!")
```

```python
def fetch_webpage_content(url):
    # adding Header to avoid blocking some websites for automatic requests
    headers = {'User-Agent': 'Mozilla/5.0'}
    response = requests.get(url, headers=headers)
    
    # analyze it using BeautifulSoup
    soup = BeautifulSoup(response.text, 'html.parser')
    
    # extract the text from the paragraphs and headings
    paragraphs = soup.find_all(['p', 'h1', 'h2', 'h3'])
    text_content = ' '.join([p.get_text(strip=True) for p in paragraphs])
    
    return text_content

target_url = "https://en.wikipedia.org/wiki/Artificial_intelligence"
Web_text = fetch_webpage_content(target_url)
```

```python
# try the function on a public website
sample_text = fetch_webpage_content("https://en.wikipedia.org/wiki/Artificial_intelligence")
print(sample_text[:500])  # print the first 500 characters to check
```

```python
def summarize_website(url):
    # 1. get the website text
    web_text = fetch_webpage_content(url)
    
    # 2. format the text into the User Prompt
    formatted_user_prompt = user_prompt_template.format(content=web_text[:200])
    
    # 3. send the request to the local Ollama
    response = ollama.chat(
        model='gemma4:e4b',
        messages=[
            {'role': 'system', 'content': system_prompt},
            {'role': 'user', 'content': formatted_user_prompt}
        ]
    )
    
    # 4. return the summary
    return response['message']['content']

# --- try the summary ---
target_url = "https://en.wikipedia.org/wiki/Artificial_intelligence"
summary = summarize_website(target_url)

print("--- summary result ---")
print(summary)
```
