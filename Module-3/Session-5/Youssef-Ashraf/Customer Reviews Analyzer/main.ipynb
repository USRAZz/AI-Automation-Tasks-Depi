```python
import ollama
import pandas as pd
```

```python
MODEL = "phi3:3.8b"
```

```python
# Read reviews from Excel
df = pd.read_excel("reviews.xlsx")
```

```python
# Get reviews from the "Review" column
reviews = df["Review"].dropna().tolist()
```

```python
# Convert reviews into text
reviews_text = "\n".join(
    f"{i + 1}. {review}"
    for i, review in enumerate(reviews)
)
```

```python
# Create the AI prompt
prompt = f"""
You are a customer feedback analyst.

Analyze these customer reviews and give me a simple summary.

Please provide:

1. Positive Points
- What customers liked about the product or service.

2. Negative Points
- What customers complained about or did not like.

3. Overall Sentiment
- Say if the reviews are Mostly Positive, Mostly Negative, or Mixed.

Customer Reviews:
{reviews_text}

Keep the answer short, clear, and easy to understand.
Use simple bullet points.
"""
```

```python
# Send reviews to Ollama
response = ollama.chat(
    model=MODEL,
    messages=[
        {
            "role": "user",
            "content": prompt
        }
    ]
)
```

```python
# Print AI analysis
print(response["message"]["content"])
```
